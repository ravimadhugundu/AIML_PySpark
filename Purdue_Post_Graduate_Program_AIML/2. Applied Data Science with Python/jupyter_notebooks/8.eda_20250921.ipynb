# EDA
- EDA stands for Exploratory Data Analysis
- It's a method of analyzing and summarizing data sets before making any assumptions or sometimes performing data cleansing.
- Advantages: 
    - identify patterns - getting familiar with the data
    - spot anomalies
    - data quality inspection
    - testing hypotheses
- Identifying data issues. During EDA, you may encounter issues like inconsistencies, duplicate records, or irrelevant features. Detecting these early on saves time and resources, ensuring you're not working with faulty data when building models.
- Defining the project objective is essential:
    - What problems are we trying to solve?
    - What are the KPIs (key performance indicators) for success?
    - What insights or outcomes do stakeholders expect?
    - What data elements I need or have accessible to solve for the problem?
    - The proper solution for the problem:
        - Statistical Analysis
        - Traditional Machine Learning
        - Advanced techniques like Deep Learning
        - Dashboard
    - If applicable: legal, privacy, and ethical considerations
- EDA Techniques
    - **Descriptive Statistical Measures**
        - Measures of Central Tendency: e.g. mean, mode, etc..
        - Measures of Dispersion: e.g. Stand Dev
        - Measures of Shape: e.g. Skewness
        - Measures of Relationship: e.g. correlation
    - **Non-Graphical Analysis**
        - Data quality checks:
            - missing values
            - appropriate data types
            - outliers
            - duplicates
        - Aggregation: e.g.
            - `groupby()`
            - `pivot_table()`
    - **Graphical Analysis** 
        - using data visualization techniques
        - popular Python libraries: `pandas`, `matplotlib`, `seaborn`, `plotly`, and `bokeh`
        - chart examples:
            - Histograms
            - Box and whisker
            - Scatter plots
            - Bar charts
            - Pie charts
            - Line plots

### Correlation Analysis

- Pearson correlation measures the linear relationship between 2 **continuous, numerical** columns
- The coefficient value ranges between -1 and 1
    - Close to +1: positive linear correlation
    - Close to -1: negative linear correlation
    - Close to 0: low or no linear correlation

```python
corr_matrix = df.corr(numeric_only=True)
corr_matrix
```

```python
plt.figure(figsize=(12,10))

sns.heatmap(corr_matrix)
```

- We can see that we have a mix of columns with zero, positive, and negative correlation 
- For prediction: 
    - The goal is to focus on the columns that are highly correlated (negative or positive) with out predicted value (SalePrice)
    - There's a lot of noise. We can summarize the heatmap further by focusing on the lane for SalePrice
- For multicollinearity analysis:
    - The goal is to focus on the correlation between the columns without the target

```python
corr_saleprice = corr_matrix[['SalePrice']].drop(index='SalePrice').sort_values(by='SalePrice', ascending=False, key=abs)
plt.figure(figsize=(2,7))
sns.heatmap(corr_saleprice, annot=True)
```

What if we want to look at columns with high correlation with SalePrice and their correlation with each other?

```python
threshold = 0.55 #we need to make this threshold work for both negative and positive

corr_saleprice = corr_matrix[['SalePrice']]
corr_saleprice_cols = corr_saleprice.loc[abs(corr_saleprice['SalePrice'])>threshold].index
corr_saleprice_cols
```

```python
plt.figure(figsize=(7,6))
high_corr_matrix = df[corr_saleprice_cols].corr()

sns.heatmap(high_corr_matrix, annot=True, cmap='viridis')
```

- High positive correlation between SalePrice and OveralQual
- Also GarageCars and GarageArea (makes sense as the bigger the garage the more cars it can fit)

Swarmplots

```python
top_5_list = df['Neighborhood'].value_counts().nlargest(5).index
```

```python
plt.style.use('dark_background')
```

```python
import matplotlib.ticker as mticker
def format_k(x, pos):
    """Formats a value with 'k' for thousands."""
    return f'{int(x/1000)}k'

df_top5_neighborhoods = df[df['Neighborhood'].isin(top_5_list)]

plt.figure(figsize=(14,7), dpi=300)
ax = sns.swarmplot(data=df_top5_neighborhoods,x='Neighborhood',y='SalePrice', size=3.5, hue='OverallQual', palette='tab20', dodge=True)
plt.yticks(range(0,df['SalePrice'].max(),50000))
ax.yaxis.set_major_formatter(mticker.FuncFormatter(format_k))
#ax.set_facecolor('black')
ax.grid(True, color='darkgray', linestyle=':')
;
```

- The higher the quality, the higher the price
- Wider cluster = more houses with similar prices
- Somerst has the highest quality out of other neighborhoods
- NAmes is almost all average quality houses and below
